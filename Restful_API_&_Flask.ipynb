1. How do you create a basic Flask application?



````python
from flask import Flask
app=Flask(__name__)

@app.route("/")
def home():
  return "Welcome"

if __name__=="__main__":
  app.run(debug=True)
````

2. How do you serve static files like images or CSS in Flask?


````python
from flask import Flask,render_template
app=Flask(__name__)

@app.route("/")
def home():
  return app.render_template("home.html") #home.html is a static file

if __name__=="__main__":
  app.run()
````

5. How can you generate URLs for routes in Flask using `url_for()`?


````python
from flask import Flask, url_for
app=Flask(__name__)

@app.route("/")
def home():
  return "welcome home!!!"
@app.route("/link")
def link():
  return f"this is to home {url_for('home')} "

if __name__=="__main__":
  app.run()
````

6. How do you handle forms in Flask?


````python
from flask import Flask,request
app=Flask(__name__)

@app.route("/submit", methods=["POST"])
def submit():
  name=request.form.get("name")
  return f"Hello {name}!!!"
````

7. How can you validate form data in Flask?


````python
from flask import Flask,request
app=Flask(__name__)

@app.route("/submit", methods=["POST"])
def submit():
  if not request.form.get("name"):
    return "name not valid",400
  return f"Hello {request.form.get('name')}"
````

11. How do you structure a Flask app using Blueprints?


````python
#auth/routes.py
from flask import Blueprint

auth_bp=Blueprint("auth",__name__, "/auth")

@auth_bp.route("/login")
def login():
  return "login page"

@auth_bp.route("logout")
def logout():
  return "logged out"
````

````python
#main.py
from flask import Flask
from auth.routes import auth_bp

app=Flask(__name__)

app.register_blueprint(auth_bp)

if __name__=="__main__":
  app.run()
````

12. How do you define a custom Jinja filter in Flask?


````python
from flask import Flask

app=Flask(__name__)

def rever(s):
  return s[::-1]

app.jinja_env.filters['reverse']=rever

@app.route("/")
def home():
  return '''{{ "hello" | reverse}}'''
````
